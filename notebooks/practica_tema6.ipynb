## Ejercicio 3 — Medición de eficiencia

```python
import random
import time
import pandas as pd

def medir_busqueda(func, estructura, claves, repeticiones=5):
    """Tiempo promedio (ms) de buscar todas las claves."""
    tiempos = []
    for _ in range(repeticiones):
        t0 = time.perf_counter()
        for c in claves:
            func(estructura, c)
        t1 = time.perf_counter()
        tiempos.append((t1 - t0) * 1000)
    return sum(tiempos) / len(tiempos)


N = 10_000
universo = list(range(N * 2))
random.shuffle(universo)
datos = universo[:N]
ordenados = sorted(datos)
claves_ok = random.sample(datos, 200)
claves_miss = random.sample([x for x in universo if x not in set(datos)], 200)
claves = claves_ok + claves_miss

# Preparar hash
tabla = TablaHash(m=N * 2)
for x in datos:
    tabla.insertar(x, x)

def buscar_hash(t, c):
    return t.buscar(c)

resultados = [
    {"Método": "Secuencial",
     "Tiempo (ms)": round(medir_busqueda(busqueda_secuencial, datos, claves), 3)},
    {"Método": "Binaria",
     "Tiempo (ms)": round(medir_busqueda(busqueda_binaria, ordenados, claves), 3)},
    {"Método": "Hash",
     "Tiempo (ms)": round(medir_busqueda(buscar_hash, tabla, claves), 3)},
]

df = pd.DataFrame(resultados)
df
```
